### Activity 5: Classes and Inheritance


```python
class MyClass(object):
    common = 10
    def __init__(self):
        self.myvariable = 3
    def myfunction(self, arg1, arg2):
        return self.myvariable

classinstance = MyClass()
print("classinstance.myfunction(1, 2):", classinstance.myfunction(1, 2))
classinstance2 = MyClass()
print("classinstance.common:", classinstance.common)
MyClass.common = 30
print("After MyClass.common = 30:", classinstance.common, classinstance2.common)
classinstance.common = 10
print("Instance override:", classinstance.common, classinstance2.common)

class OtherClass(MyClass):
    def __init__(self, arg1):
        super().__init__()
        print("OtherClass initialized with:", arg1)

sub_inst = OtherClass("hello")
print("Subclass called inherited function:", sub_inst.myfunction(1, 2))
```

### Graded Lab Task 2: Functions and OOP Hierarchy


```python
# Function definitions
def greet(name):
    print(f"Hello, {name}! Welcome to the ANN Lab.")

def fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def factorial(n):
    if n < 0:
        raise ValueError("Factorial not defined for negative numbers")
    res = 1
    for i in range(2, n + 1):
        res *= i
    return res

def reverse_string(s):
    return s[::-1]

# Demonstrations
greet("Abdullah")
print("Fibonacci(7):", fibonacci(7))
print("is_prime(29):", is_prime(29))
print("factorial(5):", factorial(5))
print("reverse_string('DeepLearning'):", reverse_string("DeepLearning"))

# Class Definitions
class Person:
    def __init__(self, name, age, email):
        self.name = name
        self.age = age
        self.email = email
    
    def introduce(self):
        print(f"Hi, I'm {self.name}, {self.age} years old. Contact: {self.email}")

class Student(Person):
    def __init__(self, name, age, email, student_id):
        super().__init__(name, age, email)
        self.student_id = student_id
    
    def study(self):
        print(f"Student {self.name} (ID: {self.student_id}) is currently studying Neural Networks.")

class Course:
    def __init__(self, course_name):
        self.course_name = course_name
        self.students = []
    
    def add_student(self, student):
        self.students.append(student)
        print(f"Enrolled {student.name} in {self.course_name}")
    
    def list_students(self):
        print(f"Enrolled students in {self.course_name}:")
        for s in self.students:
            print(f" - {s.name} ({s.student_id})")

# Combining Classes
course = Course("Introduction to Python & ANN")
s1 = Student("Alice Smith", 21, "alice@example.com", "FA23-BAI-001")
s2 = Student("Bob Johnson", 22, "bob@example.com", "FA23-BAI-002")
s3 = Student("Abdullah Ghaffar", 21, "abdullah@example.com", "FA23-BAI-059")

course.add_student(s1)
course.add_student(s2)
course.add_student(s3)

course.list_students()
s1.introduce()
s3.introduce()
s3.study()
```

### Graded Lab Task 3: Advanced Pickle Object Persistence


```python
import pickle

# 1. Explanation of Pickling:
# Pickling is the process of converting a Python object hierarchy into a byte stream
# for persistent storage or transmission over a network, and unpickling is the inverse.

# 2. Basic Pickling and Unpickling of Dicts
students_data = [
    {"name": "Alice", "age": 20, "grade": "A"},
    {"name": "Bob", "age": 22, "grade": "B"},
    {"name": "Charlie", "age": 19, "grade": "C"}
]

with open("students.pkl", "wb") as f:
    pickle.dump(students_data, f)

def load_students(filename="students.pkl"):
    with open(filename, "rb") as f:
        return pickle.load(f)

print("Loaded student dicts:", load_students())

# 3. Working with Custom Student Class
class StudentRecord:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade
    
    def __repr__(self):
        return f"StudentRecord(name='{self.name}', age={self.age}, grade='{self.grade}')"

student_objects = [StudentRecord(s["name"], s["age"], s["grade"]) for s in students_data]

with open("students_class.pkl", "wb") as f:
    pickle.dump(student_objects, f)

def load_student_objects(filename="students_class.pkl"):
    with open(filename, "rb") as f:
        return pickle.load(f)

print("Loaded Student objects:", load_student_objects())

# 4. Advanced Operations: Add & Remove Student
def add_student(filename, new_student):
    records = load_student_objects(filename)
    records.append(new_student)
    with open(filename, "wb") as f:
        pickle.dump(records, f)
    print(f"Added {new_student.name} to {filename}")

def remove_student(filename, student_name):
    records = load_student_objects(filename)
    filtered = [s for s in records if s.name != student_name]
    with open(filename, "wb") as f:
        pickle.dump(filtered, f)
    print(f"Removed {student_name} from {filename}")

# Demonstration
add_student("students_class.pkl", StudentRecord("Abdullah", 21, "A+"))
print("After addition:", load_student_objects("students_class.pkl"))

remove_student("students_class.pkl", "Bob")
print("After removal:", load_student_objects("students_class.pkl"))
```
